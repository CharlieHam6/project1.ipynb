## Data Dictionary

The dataset has 4,600 rows and 11 columns. Each row is one layoff event at a company.

| Column | Dtype | Non-Null | Use |
|---|---|---|---|
| company | str | 4600 | Which company did the layoff. Identifies each event. |
| location | str | 4599 | HQ city, tagged "Non-U.S." for international companies. Geographic comparisons. |
| total_laid_off | float64 | 3004 | Headcount cut in the event — the main size measure. Missing in ~35% of rows because news reports don't always give a number. |
| date | str | 4600 | Date of the layoff. Needs pd.to_datetime conversion for time-based analysis. |
| percentage_laid_off | float64 | 2883 | Share of the workforce cut (0 to 1). A value of 1.0 means the company shut down entirely. Also missing in many rows. |
| industry | str | 4598 | Company's sector (Finance, Crypto, Healthcare, etc.). Main categorical grouping. |
| source | str | 4597 | URL of the news article reporting the layoff. Shows data provenance; not used in analysis. |
| stage | str | 4592 | Funding stage at the time (Seed, Series A–C, Post-IPO). Proxy for company maturity. |
| funds_raised | float64 | 4054 | Total funding the company raised, in $ millions. |
| country | str | 4598 | Country of the company. Geographic grouping. |
| date_added | str | 4600 | When the row was added to the tracker (vs. date = when the layoff happened). Bookkeeping only. |
